## Concept: Speculative Decoding

Speculative decoding uses a small draft model to generate multiple candidate tokens, then verifies them with a larger target model:
1. Draft model generates K candidate tokens
2. Target model verifies all K tokens in parallel
3. Accepted tokens are added to output

**Kyro mapping**: `src/speculative.rs::SpeculativeDecoder`

```python
import random
import time
import numpy as np

class SpecDecoder:
    def __init__(self, accept_rate=0.7, lookahead=4):
        self.accept_rate = accept_rate
        self.lookahead = lookahead
    def step(self):
        draft = [random.randint(0, 99) for _ in range(self.lookahead)]
        accepted = []
        for t in draft:
            if random.random() < self.accept_rate:
                accepted.append(t)
            else:
                break
        return accepted

spec = SpecDecoder(accept_rate=0.7, lookahead=4)
N = 100

start = time.time()
for _ in range(N):
    time.sleep(0.001)
normal_time = time.time() - start

start = time.time()
for _ in range(N):
    acc = spec.step()
    time.sleep(0.001 + 0.0002 * len(acc))
spec_time = time.time() - start

print('=== Speculative Decoding ===')
print(f'Normal: {normal_time:.4f}s')
print(f'Speculative: {spec_time:.4f}s')
print(f'Speedup: {normal_time/spec_time:.2f}x')
```
