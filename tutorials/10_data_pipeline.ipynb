## Concept: Causal LM Dataset

Training data for a causal language model consists of (input, target) pairs where:
- input = tokens[0:i]
- target = tokens[1:i+1]

In practice, rolling windows create many overlapping sequences from a corpus.

```python
import numpy as np
import random

corpus = 'the cat sat on the mat the dog ran on the grass the cat slept '
tokens = corpus.lower().split()
vocab = {w: i for i, w in enumerate(set(tokens))}

seq_len = 5
datasets = []
for i in range(len(tokens) - seq_len):
    input_tokens = tokens[i:i+seq_len]
    target_tokens = tokens[i+1:i+seq_len+1]
    datasets.append((input_tokens, target_tokens))

print(f'Corpus tokens: {len(tokens)}')
print(f'Number of training examples: {len(datasets)}')
print(f'Vocabulary size: {len(vocab)}')

batch_size = 2
random.shuffle(datasets)
batches = [datasets[i:i+batch_size] for i in range(0, len(datasets), batch_size)]
print(f'Number of batches: {len(batches)}')

for batch in batches[:2]:
    for input_t, target_t in batch:
        input_ids = [vocab[w] for w in input_t]
        target_ids = [vocab[w] for w in target_t]
        print(f'  Input: {input_t} -> IDs: {input_ids}')
        print(f'  Target: {target_t} -> IDs: {target_ids}')
    print()
```
